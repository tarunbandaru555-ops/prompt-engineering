```python
# ============================================================
# EXPERIMENT: PROMPT ENGINEERING COMPARISON
# Google Colab + Groq API
# ============================================================

# ------------------------------------------------------------
# STEP 1: Install Groq
# ------------------------------------------------------------

!pip -q install -U groq


# ------------------------------------------------------------
# STEP 2: Import libraries
# ------------------------------------------------------------

import getpass
from groq import Groq


# ------------------------------------------------------------
# STEP 3: Get GROQ API Key
# ------------------------------------------------------------
# In Google Colab:
# 1. Click the 🔑 Secrets icon on the left
# 2. Add a secret named:
#
#       GROQ_API_KEY
#
# 3. Enable "Notebook access"
#
# The code will automatically read the key.

try:
    from google.colab import userdata
    api_key = userdata.get("GROQ_API_KEY")
except Exception:
    api_key = None

# If the key was not found, ask for it
if not api_key:
    api_key = getpass.getpass("Enter your Groq API key: ")


# ------------------------------------------------------------
# STEP 4: Create Groq client
# ------------------------------------------------------------

client = Groq(api_key=api_key)


# ------------------------------------------------------------
# STEP 5: Select a CURRENT Groq model
# ------------------------------------------------------------
# Groq currently supports:
#
# openai/gpt-oss-20b
# openai/gpt-oss-120b
#
# We use GPT-OSS 20B because it is suitable for this
# simple prompt-engineering experiment.

MODEL = "openai/gpt-oss-20b"

TEMPERATURE = 0.2
MAX_TOKENS = 150


# ------------------------------------------------------------
# STEP 6: Function to send prompt
# ------------------------------------------------------------

def send_prompt(prompt):
    """
    Sends a prompt to the Groq LLM and returns the response.
    """

    response = client.chat.completions.create(
        model=MODEL,
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ],
        temperature=TEMPERATURE,
        max_tokens=MAX_TOKENS
    )

    return response.choices[0].message.content.strip()


# ------------------------------------------------------------
# STEP 7: BASELINE PROMPT
# ------------------------------------------------------------

baseline_prompt = (
    "Explain what recursion is to a beginner."
)


# ------------------------------------------------------------
# STEP 8: ENHANCED PROMPT
# ------------------------------------------------------------

enhanced_prompt = (
    "You are a computer science professor. "
    "Explain recursion to a beginner using a simple real-world analogy. "
    "Give a clear definition and one example. "
    "Keep the explanation under 80 words."
)


# ------------------------------------------------------------
# STEP 9: Send both prompts
# ------------------------------------------------------------
# Each prompt is sent separately.
# Therefore, there is NO context leakage.

print("Sending baseline prompt...")
baseline_response = send_prompt(baseline_prompt)

print("Sending enhanced prompt...")
enhanced_response = send_prompt(enhanced_prompt)


# ------------------------------------------------------------
# STEP 10: Functions for comparison
# ------------------------------------------------------------

def word_count(text):
    """Count the number of words."""
    return len(text.split())


def sentence_count(text):
    """Approximately count the number of sentences."""
    return sum(text.count(mark) for mark in [".", "!", "?"])


# Calculate statistics

baseline_words = word_count(baseline_response)
enhanced_words = word_count(enhanced_response)

baseline_sentences = sentence_count(baseline_response)
enhanced_sentences = sentence_count(enhanced_response)


# ------------------------------------------------------------
# STEP 11: Display BASELINE prompt and response
# ------------------------------------------------------------

print("\n")
print("=" * 70)
print("                    BASELINE PROMPT")
print("=" * 70)

print("\nPrompt:")
print(baseline_prompt)

print("\nResponse:")
print(baseline_response)


# ------------------------------------------------------------
# STEP 12: Display ENHANCED prompt and response
# ------------------------------------------------------------

print("\n")
print("=" * 70)
print("                    ENHANCED PROMPT")
print("=" * 70)

print("\nPrompt:")
print(enhanced_prompt)

print("\nResponse:")
print(enhanced_response)


# ------------------------------------------------------------
# STEP 13: Comparison
# ------------------------------------------------------------

print("\n")
print("=" * 70)
print("                       COMPARISON")
print("=" * 70)

print()

print(f"{'Metric':<25} {'Baseline':<20} {'Enhanced':<20}")
print("-" * 65)

print(
    f"{'Word Count':<25} "
    f"{baseline_words:<20} "
    f"{enhanced_words:<20}"
)

print(
    f"{'Sentence Count':<25} "
    f"{baseline_sentences:<20} "
    f"{enhanced_sentences:<20}"
)


# ------------------------------------------------------------
# STEP 14: Analysis
# ------------------------------------------------------------

print("\n")
print("=" * 70)
print("                         ANALYSIS")
print("=" * 70)

print("""
1. SPECIFICITY
   The baseline prompt only asks the model to explain recursion.
   The enhanced prompt gives a specific role, target audience,
   analogy, definition, example, and word limit.

2. CLARITY
   The enhanced prompt provides clearer instructions about what
   information should be included in the answer.

3. READABILITY
   The enhanced prompt asks for a simple real-world analogy,
   making the explanation more suitable for beginners.

4. OUTPUT CONTROL
   The enhanced prompt includes a specific length constraint and
   required content, producing a more focused response.

5. FAIR COMPARISON
   Both prompts use the same model and the same parameters:
   temperature = 0.2
   max_tokens = 150

   The prompts are also sent independently, so there is no
   context leakage between the two requests.

CONCLUSION
   Prompt engineering can improve the quality, clarity, and
   usefulness of language-model responses by adding role,
   context, constraints, and output-format instructions.
""")


# ------------------------------------------------------------
# STEP 15: Experiment completed
# ------------------------------------------------------------

print("=" * 70)
print("             EXPERIMENT COMPLETED SUCCESSFULLY")
print("=" * 70)
```
